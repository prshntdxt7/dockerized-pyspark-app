```python
import os
import subprocess
from datetime import datetime

dt = str(datetime.today().strftime('%Y%m%d'))

# Define the paths
images_path = '/visual-content-case-study-main/data/images.jsonl'
tags_path = '/visual-content-case-study-main/data/image_tags.jsonl'
main_images_path = '/visual-content-case-study-main/data/main_images.jsonl'
output_cdc_path = f'/visual-content-case-study-main/output/output_cdc_{dt}.jsonl'
output_snapshot_path = f'/visual-content-case-study-main/output/output_snapshot_{dt}.jsonl'
output_metrics_path = f'/visual-content-case-study-main/output/output_metrics_{dt}.jsonl'

# Construct the command
command = [
    'python', '/visual-content-case-study-main/src/process_images.py',
    '--images', images_path,
    '--tags', tags_path,
    '--main_images', main_images_path,
    '--output_cdc', output_cdc_path,
    '--output_snapshot', output_snapshot_path,
    '--output_metrics', output_metrics_path
]

try:
    # Run the command and capture output and errors
    result = subprocess.run(command, check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    print("Output:\n", result.stdout)
    print("Errors:\n", result.stderr)
except subprocess.CalledProcessError as e:
    print("Command failed with error code", e.returncode)
    print("Output:\n", e.stdout)
    print("Errors:\n", e.stderr)

print("Script execution [COMPLETED] | Output Folder: [output]")



```
